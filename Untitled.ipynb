```python
def goertzel(data,fs,tone):
    N = len(data)
    mag = 0
    
    k0 = N*tone/fs
    
    omega_I = np.cos(2*np.pi*k0/N)
    omega_Q = np.sin(2*np.pi*k0/N)
    v1 = 0
    v2 = 0
    for n in range(N):
        v  = 2*omega_I*v1 - v2 + data[n]
        v2 = v1
        v1 = v
      
    y_I = v1 - omega_I*v2
    y_Q = omega_Q*v2
      
    mag = np.sqrt(y_I**2 + y_Q**2)
    return mag
```

```python
def analyze(data,num_samples,duration,tone):
    fs=num_samples/duration
    x=np.linspace(0,duration,len(data))
    plt.figure(figsize=(20,4))
    plt.plot(x,data)
    x1=np.arange(0,duration,0.001)
    plt.vlines(x1,np.min(data),np.max(data),colors='red')
    plt.show()

    spec,freq,line=plt.magnitude_spectrum(data,fs)
    plt.show(line)

    mag=goertzel(data,fs,tone)
    print(mag)
```
